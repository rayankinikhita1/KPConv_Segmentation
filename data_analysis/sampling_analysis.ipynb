```python

import numpy as np
import matplotlib.pyplot as plt
from plyfile import PlyData
from matplotlib.colors import ListedColormap
label_id_to_name = { 0: "ground",
                                1: "wall",
                                2: "busbar",
                                3: "wire",
                                4: "tower",
                                5: "media",
                                6: "builds",
                                7: "clutter",
                                8: "small",
                                9: "large"}



# Load PLY file with RGB and scalar_classification
ply_path = '/home/nikhita_rayanki/projects/datasets/grid_subsampled_files/Elecs_1/substation1.ply'  # <-- Update this path
ply_data = PlyData.read(ply_path)
x = np.array(ply_data['vertex']['x'])
y = np.array(ply_data['vertex']['y'])
z = np.array(ply_data['vertex']['z'])
r = np.array(ply_data['vertex']['red'])
g = np.array(ply_data['vertex']['green'])
b = np.array(ply_data['vertex']['blue'])
classes = np.array(ply_data['vertex']['class']).astype(int)

unique_classes = np.unique(classes)
num_classes = len(label_id_to_name)
# Use tab20 for up to 20 classes, otherwise combine tab20b and tab20c
if num_classes <= 20:
    cmap = plt.get_cmap('tab20', num_classes)
else:
    cmap = ListedColormap(
        np.vstack([plt.get_cmap('tab20b').colors, plt.get_cmap('tab20c').colors])[:num_classes]
    )

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')
sc = ax.scatter(x, y, z, c=classes, cmap=cmap, s=0.1, vmin=0, vmax=num_classes-1)

# Only show ticks for present classes
cbar = plt.colorbar(sc, ax=ax, pad=0.1, ticks=unique_classes)
cbar.ax.set_yticklabels([label_id_to_name[i] for i in unique_classes])

plt.title('Point Cloud Visualization with RGB and Class Colors')
plt.tight_layout()
plt.show()
plt.close()
```
